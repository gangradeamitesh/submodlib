```python
import matplotlib.pyplot as plt
from submodlib import FacilityLocationVariantMutualInformationFunction as LegacyFLVMI
from submodlib.sub_modularfunctions.mutual_info_functions.FLVMI import (
    FacilityLocationVariantMutualInformation as PyFLVMI,
)

etas = [0, 0.4, 0.8, 1, 1.4, 1.8, 2.2, 2.6, 3, 10, 50, 100]

def summarize(greedy):
    # convert [ (idx, gain), ... ] → [(idx, round(gain,4)), ...] for readability
    return [(int(idx), float(gain)) for idx, gain in greedy]

def run_version(obj):
    greedy = obj.maximize(
        budget=10,
        optimizer="NaiveGreedy",
        stopIfZeroGain=False,
        stopIfNegativeGain=False,
        epsilon=0.0,
        verbose=False,
        show_progress=False,
        costs=None,
        costSensitiveGreedy=False,
    )
    return greedy

comparison = []

for eta in etas:
    # Legacy (C++ backed) version
    legacy_obj = LegacyFLVMI(
        n=46,
        num_queries=2,
        data=groundData,
        queryData=mutlipleQueryData,
        metric="euclidean",
        queryDiversityEta=eta,
    )
    legacy_greedy = run_version(legacy_obj)

    # New PyTorch version (make sure ground/query tensors match expectations)
    py_obj = PyFLVMI(
        n=46,
        num_queries=2,
        data=groundData,
        query_data=mutlipleQueryData2,
        metric="euclidean",
        queryDiversityEta=eta,
    )
    py_greedy = run_version(py_obj)

    comparison.append(
        {
            "eta": eta,
            "legacy": summarize(sorted(legacy_greedy, key=lambda x: x[1], reverse=True)),
            "python": summarize(sorted(py_greedy, key=lambda x: x[1], reverse=True)),
        }
    )

comparison
```

```python
ncols = 2
nrows = int(np.ceil(len(etas) / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(14, 4 * nrows))

for ax_idx, (eta, panel) in enumerate(zip(etas, comparison)):
    row, col = divmod(ax_idx, ncols)
    ax = axes[row, col]
    # choose which run to display; here legacy vs python on top/bottom by changing data source
    greedy = comparison[ax_idx]["legacy"]  # or ["python"]
    greedy_xs = [groundxs[idx] for idx, _ in greedy]
    greedy_ys = [groundys[idx] for idx, _ in greedy]

    ax.scatter(groundxs, groundys, s=50, facecolors="none", edgecolors="black", label="Images")
    ax.scatter(multiplequeryxs, multiplequeryys, s=50, color="green", label="Queries")
    ax.scatter(greedy_xs, greedy_ys, s=50, color="C0", label="Greedy Set")
    for label, (idx, _) in enumerate(greedy):
        ax.annotate(label, (groundxs[idx], groundys[idx]), (groundxs[idx] + 0.1, groundys[idx] + 0.1))
    ax.set_title(f"η = {eta}")

# hide unused subplot if len(etas) is odd
if len(etas) % ncols != 0:
    axes[-1, -1].axis("off")

plt.tight_layout()
```
